## Plots

```python
names = time_series.getSpeciesNames()
names_lower = np.array([name.lower() for name in names])
names = names[np.argsort(names_lower)]
```

### Convergence over all species

```python
def calculateAbsErr(marginal_distribution, ord=None):
    difference = np.zeros((len(names), len(time)))
    for i, name in enumerate(names):
        difference[i, :] = marginal_distribution[name] - reference_r60[0][name]
    abs_err = np.linalg.norm(difference, axis=0, ord=ord)
    return abs_err
```

```python
def calculateAbsErr_r5(marginal_distribution, ord=None):
    difference = np.zeros((len(names), len(time)))
    for i, name in enumerate(names):
        difference[i, :] = marginal_distribution[name] - moments_pb_r5_e[0][name]
    abs_err = np.linalg.norm(difference, axis=0, ord=ord)
    return abs_err
```

```python
def calculateAbsErr_r10(marginal_distribution, ord=None):
    difference = np.zeros((len(names), len(time)))
    for i, name in enumerate(names):
        difference[i, :] = marginal_distribution[name] - moments_pb_r10_e[0][name]
    abs_err = np.linalg.norm(difference, axis=0, ord=ord)
    return abs_err
```

```python
def calculateAbsErr_r20(marginal_distribution, ord=None):
    difference = np.zeros((len(names), len(time)))
    for i, name in enumerate(names):
        difference[i, :] = marginal_distribution[name] - moments_pb_r20_e[0][name]
    abs_err = np.linalg.norm(difference, axis=0, ord=ord)
    return abs_err
```

### Error against matrix solution (RK4)

```python
ord = np.inf

fig, axs = plt.subplots(1, 3, figsize=(7, 2.3))

axs[0].plot(time, calculateAbsErr_r5(moments_pb_r5_5[0], ord=ord), ":", color=colors[0], label="$r=5,5$")
axs[0].plot(time, calculateAbsErr_r5(moments_pb_r5_10[0], ord=ord), "-.", color=colors[0], label="$r=5,10$")
axs[0].plot(time, calculateAbsErr_r5(moments_pb_r5_20[0], ord=ord), "--", color=colors[0], label="$r=5,20$")
axs[0].plot(time, calculateAbsErr_r5(moments_pb_r5_30[0], ord=ord), "-.", color=colors[0], label="$r=5,20$")

axs[1].plot(time, calculateAbsErr_r10(moments_pb_r10_10[0], ord=ord), "-.", color=colors[2], label="$r=10,10$")
axs[1].plot(time, calculateAbsErr_r10(moments_pb_r10_20[0], ord=ord), "--", color=colors[2], label="$r=10,20$")
axs[1].plot(time, calculateAbsErr_r10(moments_pb_r10_30[0], ord=ord), "-", color=colors[2], label="$r=10,30$")

axs[2].plot(time, calculateAbsErr_r20(moments_pb_r20_20[0], ord=ord), ":", color=colors[1], label="$r=20,20$")
axs[2].plot(time, calculateAbsErr_r20(moments_pb_r20_30[0], ord=ord), "-.", color=colors[1], label="$r=20,30$")
axs[2].plot(time, calculateAbsErr_r20(moments_pb_r20_40[0], ord=ord), "-.", color=colors[1], label="$r=20,40$")
axs[2].plot(time, calculateAbsErr_r20(moments_pb_r20_50[0], ord=ord), "-", color=colors[1], label="$r=20,40$")

# fig.legend([l0, l1, l2], ["best", "reasonable", "worst"], ncols=3, loc="center", bbox_to_anchor=(0.5, 1.01))
plt.setp(axs, xlabel="$t$", ylim=[2e-3, 3e-2], xticks=[0, 5, 10, 15, 20], yscale="log")
# axs[0].set_ylabel("$\| \langle x_i \\rangle (t) - \langle x_{{i,\mathrm{{ref}}}} \\rangle (t) \|_{{\infty,i}}$")

axs[0].set_title("$r=5$")
axs[1].set_title("$r=10$")
axs[2].set_title("$r=20$")

axs[1].set_yticklabels([])

axs[2].yaxis.tick_right()
axs[2].yaxis.set_ticks_position("both")
axs[2].yaxis.set_label_position("right")

plt.subplots_adjust(wspace=0.03)
# plt.savefig("plots/pancreatic_moments_err_infty_separate.pdf", bbox_inches="tight")
```

### Error against reference solution (RK4)

```python
ord = np.inf

fig, axs = plt.subplots(1, 3, figsize=(7, 2.3))

absErr = np.zeros((3, 4, len(time)))

absErr[0,:] = [calculateAbsErr(moments_pb_r5_5[0], ord=ord), 
             calculateAbsErr(moments_pb_r5_10[0], ord=ord), 
             calculateAbsErr(moments_pb_r5_20[0], ord=ord), 
             calculateAbsErr(moments_pb_r5_30[0], ord=ord)]

absErr[1,:] = [calculateAbsErr(moments_pb_r10_10[0], ord=ord), 
              calculateAbsErr(moments_pb_r10_20[0], ord=ord), 
              calculateAbsErr(moments_pb_r10_30[0], ord=ord),
              calculateAbsErr(moments_pb_r10_40[0], ord=ord)]

absErr[2,:] = [calculateAbsErr(moments_pb_r20_20[0], ord=ord), 
              calculateAbsErr(moments_pb_r20_30[0], ord=ord), 
              calculateAbsErr(moments_pb_r20_40[0], ord=ord),
              calculateAbsErr(moments_pb_r20_50[0], ord=ord)]

axs[0].plot(time, absErr[0][0], ":", color=colors[0])
axs[0].plot(time, absErr[0][1], "-.", color=colors[0])
axs[0].plot(time, absErr[0][2], "--", color=colors[0])
axs[0].plot(time, absErr[0][3], "-", color=colors[0])
l0 = axs[0].plot(time, calculateAbsErr(moments_pb_r5[0], ord=ord), "k:")

axs[1].plot(time, absErr[1][0], ":", color=colors[0])
axs[1].plot(time, absErr[1][1], "-.", color=colors[0])
axs[1].plot(time, absErr[1][2], "--", color=colors[0])
axs[1].plot(time, absErr[1][3], "-", color=colors[0])
axs[1].plot(time, calculateAbsErr(moments_pb_r10[0], ord=ord), "k:")

axs[2].plot(time, absErr[2][0], ":", color=colors[0])
axs[2].plot(time, absErr[2][1], "-.", color=colors[0])
axs[2].plot(time, absErr[2][2], "--", color=colors[0])
axs[2].plot(time, absErr[2][3], "-", color=colors[0])
axs[2].plot(time, calculateAbsErr(moments_pb_r20[0], ord=ord), "k:", label="one-level part.")
axs[2].legend()

plt.setp(axs, xlabel="$t$", ylim=[1e-3, 4e-2], xticks=[0, 5, 10, 15, 20], yscale="log")
axs[0].set_ylabel("$\| \langle x_i \\rangle (t) - \langle x_{{i,\mathrm{{ref}}}} \\rangle (t) \|_{{\infty,i}}$")

axs[0].set_title("$r=5$")
axs[1].set_title("$r=10$")
axs[2].set_title("$r=20$")

rank_axs0 = [5, 10, 20, 30]
incr_axs0 = [0, 0, 2, -2]
for j in range(4):
    axs[0].annotate(
        text="${}$".format(rank_axs0[j]),
        xy=(time[-1], absErr[0][j][-1]),
        xytext=(2, incr_axs0[j]),
        textcoords="offset points",
        va="center",
        ha="right",
        fontsize="x-small",
        bbox=dict(boxstyle="square,pad=0", fc="white", ec="none")
    )

incr_axs12 = [[0, 0, 1, -1], [0, 2.25, 0.25, -2.5]]
for i in range(1,3):
    for j in range(4):
        axs[i].annotate(
            text="${}$".format((j + i) * 10),
            xy=(time[-1], absErr[i][j][-1]),
            xytext=(2, incr_axs12[i-1][j]),
            textcoords="offset points",
            va="center",
            ha="right",
            fontsize="x-small",
            bbox=dict(boxstyle="square,pad=0", fc="white", ec="none")
        )

axs[1].set_yticklabels([])

axs[2].yaxis.tick_right()
axs[2].yaxis.set_ticks_position("both")
axs[2].yaxis.set_label_position("right")

plt.subplots_adjust(wspace=0.03)
plt.savefig("plots/pancreatic_moments_convergence_RK4.pdf", bbox_inches="tight")
```
